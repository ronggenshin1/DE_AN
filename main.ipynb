```python
# cif_to_qe.py

from ase.io import read, write

# 1. Đọc file cấu trúc tinh thể .cif (thay tên file .cif thực tế của bạn vào đây)
cif_file = "structure.cif"
atoms = read(cif_file)

# 2. Cấu hình các thông số tính toán cho Quantum ESPRESSO (&CONTROL, &SYSTEM, &ELECTRONS)
input_data = {
    "control": {
        "calculation": "scf",
        "restart_mode": "from_scratch",
        "prefix": "calc_system",
        "pseudo_dir": "./pseudos",  # Đường dẫn tới thư mục chứa file .upf
        "outdir": "./out",
        "tstress": True,
        "tprnfor": True,
    },
    "system": {
        "ecutwfc": 45.0,  # Wavefunction cutoff energy (Ry)
        "ecutrho": 360.0,  # Charge density cutoff energy (Ry)
        "occupations": "smearing",
        "smearing": "gaussian",
        "degauss": 0.01,
    },
    "electrons": {
        "conv_thr": 1.0e-8,
        "mixing_beta": 0.7,
    },
}

# 3. Khai báo file Pseudopotential (.upf) tương ứng với từng nguyên tố trong file .cif
# Điền tên file .upf có trong thư mục pseudos của bạn vào đây
pseudopotentials = {
    "Fe": "Fe.paw.pbe.z_16.ld1.psl.v0.2.1.upf",
    "O": "O.paw.pbe.z_6.ld1.psl.v0.1.upf",
}

# 4. Xuất ra file input (.in) cho Quantum ESPRESSO
output_in_file = "pw.scf.in"

write(
    output_in_file,
    atoms,
    format="espresso-in",
    input_data=input_data,
    pseudopotentials=pseudopotentials,
    kpts=(4, 4, 4),  # Lưới K-points (Monkhorst-Pack grid)
    koffset=(0, 0, 0),
)

print(f"-> Đã chuyển đổi thành công từ '{cif_file}' sang '{output_in_file}'!")
```
